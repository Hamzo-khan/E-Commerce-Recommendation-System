```python
import streamlit as st
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# Load Dataset
df = pd.read_csv("Recommendation_System_Big_Dataset.csv")

# Create User-Product Matrix
user_product_matrix = df.pivot_table(
    index="User_ID",
    columns="Product_Name",
    aggfunc="size",
    fill_value=0
)

# Product Similarity
product_matrix = user_product_matrix.T

product_similarity = cosine_similarity(product_matrix)

product_similarity_df = pd.DataFrame(
    product_similarity,
    index=product_matrix.index,
    columns=product_matrix.index
)

# Recommendation Function
def recommend(product_name):
    similar_products = (
        product_similarity_df.loc[product_name]
        .sort_values(ascending=False)
        .iloc[1:6]
    )
    return similar_products.index.tolist()

# ---------------- GUI ----------------

st.title("🛒 Product Recommendation System")

product = st.selectbox(
    "Select Product",
    sorted(df["Product_Name"].unique())
)

if st.button("Recommend"):

    st.subheader("Recommended Products")

    recommendations = recommend(product)

    for i, item in enumerate(recommendations,1):
        st.write(f"{i}. {item}")
```

```python
import gradio as gr

def recommend(product_name):
    similar_products = (
        product_similarity_df.loc[product_name]
        .sort_values(ascending=False)
        .iloc[1:6]
    )

    return "\n".join(similar_products.index.tolist())


demo = gr.Interface(
    fn=recommend,
    inputs=gr.Dropdown(
        choices=sorted(df["Product_Name"].unique()),
        label="Select Product"
    ),
    outputs=gr.Textbox(label="Recommended Products"),
    title="🛒 Product Recommendation System",
    description="Select a product to get recommendations."
)

demo.launch()
```

```python
import gradio as gr

# Recommendation Function
def recommend(product_name):

    similar_products = (
        product_similarity_df.loc[product_name]
        .sort_values(ascending=False)
        .iloc[1:6]
    )

    return "\n".join(
        [f"⭐ {i+1}. {product}" for i, product in enumerate(similar_products.index)]
    )


theme = gr.themes.Soft(
    primary_hue="blue",
    secondary_hue="cyan",
    neutral_hue="slate"
)

with gr.Blocks(theme=theme, title="Product Recommendation System") as demo:

    gr.Markdown(
        """
        # 🛒 Product Recommendation System

        ### Find Similar Products Instantly

        Select a product and click **Get Recommendations**.
        """
    )

    with gr.Row():

        product = gr.Dropdown(
            choices=sorted(df["Product_Name"].unique()),
            label="📦 Select Product",
            value="Gaming Mouse"
        )

    btn = gr.Button(
        "🚀 Get Recommendations",
        variant="primary"
    )

    output = gr.Textbox(
        label="🎯 Recommended Products",
        lines=8
    )

    btn.click(
        recommend,
        inputs=product,
        outputs=output
    )

demo.launch()
```

```python
import gradio as gr
import random

# Random Placeholder Images
images = [
    "https://picsum.photos/500/350?random=1",
    "https://picsum.photos/500/350?random=2",
    "https://picsum.photos/500/350?random=3",
    "https://picsum.photos/500/350?random=4",
    "https://picsum.photos/500/350?random=5",
    "https://picsum.photos/500/350?random=6"
]

def recommend_ui(product_name):

    similar_products = (
        product_similarity_df.loc[product_name]
        .sort_values(ascending=False)
        .iloc[1:6]
    )

    img = random.choice(images)

    recommendations = ""

    for i, p in enumerate(similar_products.index, 1):
        recommendations += f"{i}. {p}\n"

    info = f"""
📦 Product : {product_name}

⭐ Recommendation Results

{recommendations}
"""

    return img, info


theme = gr.themes.Soft(
    primary_hue="blue",
    secondary_hue="cyan"
)

with gr.Blocks(theme=theme) as demo:

    gr.Markdown(
        """
# 🛒 Product Recommendation System

### AI Based Product Recommendation

Select a product and click **Get Recommendations**
"""
    )

    with gr.Row():

        product = gr.Dropdown(
            choices=sorted(df["Product_Name"].unique()),
            label="📦 Select Product"
        )

    btn = gr.Button(
        "🚀 Get Recommendations",
        variant="primary"
    )

    with gr.Row():

        image = gr.Image(
            label="🖼️ Product Image",
            height=300
        )

        output = gr.Textbox(
            label="🎯 Recommendations",
            lines=12
        )

    btn.click(
        recommend_ui,
        inputs=product,
        outputs=[image, output]
    )

demo.launch()
```
